```python
import math
import torch
torch.manual_seed(0)

batch_size = 1
seq_len    = 4
d_model    = 8
d_k        = 8

X = torch.randn(batch_size, seq_len, d_model)

W_Q = torch.nn.Linear(d_model, d_k, bias=False)
W_K = torch.nn.Linear(d_model, d_k, bias=False)
W_V = torch.nn.Linear(d_model, d_k, bias=False)

def scaled_dot_product_attention(X):
    Q = W_Q(X)                            # (B, L, d_k)
    K = W_K(X)                            # (B, L, d_k)
    V = W_V(X)                            # (B, L, d_k)

    # (B, L, L): for each token, similarity to every token
    attn_logits = Q @ K.transpose(-2, -1) / math.sqrt(d_k)

    # Softmax on last dim -> rows sum to 1
    attn_weights = torch.softmax(attn_logits, dim=-1)

    # Weighted sum of values
    out = attn_weights @ V                # (B, L, d_k)
    return out, attn_weights

out, attn_weights = scaled_dot_product_attention(X)

print("X.shape:", X.shape)
print("attn_weights.shape:", attn_weights.shape)
print("out.shape:", out.shape)
print("Row sums of attention (should be ~1):")
print(attn_weights[0].sum(dim=-1))
```
