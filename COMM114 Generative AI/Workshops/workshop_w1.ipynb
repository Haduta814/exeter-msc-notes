(1) Compute entropy of a binary random variable with probability mess function {0.8, 0.2} by using numpy package.

```python
# Compute entropy by using numpy package
import numpy as np


p = np.array([0.8, 0.2])
logp = np.log(p)
entropy1 = np.sum(-p*logp)
print(entropy1)
```

(2) Compute entropy of a binary random variable with probability mess function {0.8, 0.2} by using pytorch.

```python
# Compute entropy by using pytorch

import torch

from torch.distributions import Categorical

p_tensor = torch.Tensor([0.8, 0.2])
entropy2 = Categorical(probs = p_tensor).entropy()
print(entropy2)

```

(3) Compute KL divergence of two probabilites P and Q by using pytorch, where P={0.8, 0.2} and Q={0.4, 0.6}

```python
import torch.nn.functional as F

# this is the same example in wiki
P = torch.Tensor([0.8, 0.2])
Q = torch.Tensor([0.4, 0.6])

print((P * (P / Q).log()).sum())

print(F.kl_div(Q.log(), P, None, None, 'sum'))
```
