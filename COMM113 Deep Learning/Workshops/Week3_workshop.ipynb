```python
import torch
import torch.nn as nn
import numpy as np

# ============================================================
# TASK 1: Training with PyTorch (One Iteration)
# ============================================================

# Create sample input data (batch_size=5, features=4)
X_torch = torch.tensor([
    [1.0, 2.0, 3.0, 4.0],
    [2.0, 3.0, 4.0, 5.0],
    [3.0, 4.0, 5.0, 6.0],
    [4.0, 5.0, 6.0, 7.0],
    [5.0, 6.0, 7.0, 8.0]
], dtype=torch.float32)

# Create sample labels (batch_size=5)
y_torch = torch.tensor([0, 1, 2, 1, 0], dtype=torch.long)
```

```python
# Network parameters
N = 4  # Input features
H = 8  # Hidden layer neurons
K = 3  # Output classes

# Define the network with ReLU and Softmax
class SimpleNetwork(nn.Module):
    def __init__(self, N, H, K):
        super(SimpleNetwork, self).__init__()
        self.fc1 = nn.Linear(N, H)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(H, K)
        self.softmax = nn.Softmax(dim=1)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        x = self.softmax(x)
        return x

# Set manual seed for reproducibility
# torch.manual_seed(42)
# Create model
model = SimpleNetwork(N, H, K)

print("\n--- INITIAL WEIGHTS ---")
print(f"\nLayer 1 (fc1) Weight shape: {model.fc1.weight.shape}")
print(f"Layer 1 (fc1) Weight:\n{model.fc1.weight.data}")
print(f"\nLayer 1 (fc1) Bias shape: {model.fc1.bias.shape}")
print(f"Layer 1 (fc1) Bias:\n{model.fc1.bias.data}")

print(f"\nLayer 2 (fc2) Weight shape: {model.fc2.weight.shape}")
print(f"Layer 2 (fc2) Weight:\n{model.fc2.weight.data}")
print(f"\nLayer 2 (fc2) Bias shape: {model.fc2.bias.shape}")
print(f"Layer 2 (fc2) Bias:\n{model.fc2.bias.data}")
```

```python
# Define loss function and optimizer
# Note: Using NLLLoss with log_softmax for numerical stability
criterion = nn.NLLLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# Forward pass
output = model(X_torch)
print(f"\n--- FORWARD PASS ---")
print(f"Model output shape: {output.shape}")
print(f"Model output (after softmax):\n{output}")

# Calculate loss (convert softmax to log for NLLLoss)
loss = criterion(torch.log(output + 1e-8), y_torch)
print(f"\nLoss: {loss.item():.6f}")

# Backward pass
optimizer.zero_grad()
loss.backward()

print("\n--- GRADIENTS ---")
print(f"\nLayer 1 (fc1) Weight gradient:\n{model.fc1.weight.grad}")
print(f"\nLayer 1 (fc1) Bias gradient:\n{model.fc1.bias.grad}")
print(f"\nLayer 2 (fc2) Weight gradient:\n{model.fc2.weight.grad}")
print(f"\nLayer 2 (fc2) Bias gradient:\n{model.fc2.bias.grad}")

# Update weights
optimizer.step()

print("\n--- UPDATED WEIGHTS ---")
print(f"\nLayer 1 (fc1) Weight:\n{model.fc1.weight.data}")
print(f"\nLayer 1 (fc1) Bias:\n{model.fc1.bias.data}")
print(f"\nLayer 2 (fc2) Weight:\n{model.fc2.weight.data}")
print(f"\nLayer 2 (fc2) Bias:\n{model.fc2.bias.data}")
```

```python
# ============================================================
# TASK 2: Training with NumPy (ONE ITERATION) - ReLU + Softmax
# ============================================================

# Create sample input data
X_numpy = np.array([
    [1.0, 2.0, 3.0, 4.0],
    [2.0, 3.0, 4.0, 5.0],
    [3.0, 4.0, 5.0, 6.0],
    [4.0, 5.0, 6.0, 7.0],
    [5.0, 6.0, 7.0, 8.0]
], dtype=np.float32)

# Create sample labels (one-hot encoded for cross-entropy)
y_numpy = np.array([0, 1, 2, 1, 0], dtype=np.int64)
y_onehot = np.zeros((5, 3))
y_onehot[np.arange(5), y_numpy] = 1

# Initialize weights with same seed
#np.random.seed(42)
W1 = np.random.randn(N, H).astype(np.float32) * 0.1
b1 = np.random.randn(H).astype(np.float32) * 0.1
W2 = np.random.randn(H, K).astype(np.float32) * 0.1
b2 = np.random.randn(K).astype(np.float32) * 0.1

print("\n--- INITIAL WEIGHTS ---")
print(f"\nLayer 1 Weight shape: {W1.shape}")
print(f"Layer 1 Weight:\n{W1}")
print(f"\nLayer 1 Bias shape: {b1.shape}")
print(f"Layer 1 Bias:\n{b1}")

print(f"\nLayer 2 Weight shape: {W2.shape}")
print(f"Layer 2 Weight:\n{W2}")
print(f"\nLayer 2 Bias shape: {b2.shape}")
print(f"Layer 2 Bias:\n{b2}")
```

```python
# Define activation functions
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(np.float32)

def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)

def cross_entropy_loss(predictions, targets):
    n_samples = predictions.shape[0]
    return -np.sum(targets * np.log(predictions + 1e-8)) / n_samples

# Forward propagation
z1 = X_numpy @ W1 + b1  # Linear transformation
a1 = relu(z1)           # ReLU activation
z2 = a1 @ W2 + b2       # Linear transformation
a2 = softmax(z2)        # Softmax activation

print(f"\n--- FORWARD PASS ---")
print(f"Hidden layer output (after ReLU) shape: {a1.shape}")
print(f"Hidden layer output (after ReLU):\n{a1}")
print(f"Final output (after softmax) shape: {a2.shape}")
print(f"Model output (after softmax):\n{a2}")
```

```python
# Calculate loss
loss = cross_entropy_loss(a2, y_onehot)
print(f"\nLoss: {loss:.6f}")

# Backward pass
learning_rate = 0.01
n_samples = X_numpy.shape[0]

# Output layer gradients
dz2 = (a2 - y_onehot) / n_samples  # Gradient of loss w.r.t. z2
dW2 = a1.T @ dz2                    # Gradient w.r.t. W2
db2 = np.sum(dz2, axis=0)           # Gradient w.r.t. b2

# Hidden layer gradients
da1 = dz2 @ W2.T                    # Gradient w.r.t. a1
dz1 = da1 * relu_derivative(z1)     # Gradient w.r.t. z1 (ReLU derivative)
dW1 = X_numpy.T @ dz1               # Gradient w.r.t. W1
db1 = np.sum(dz1, axis=0)           # Gradient w.r.t. b1

print("\n--- GRADIENTS ---")
print(f"\nLayer 1 Weight gradient:\n{dW1}")
print(f"\nLayer 1 Bias gradient:\n{db1}")
print(f"\nLayer 2 Weight gradient:\n{dW2}")
print(f"\nLayer 2 Bias gradient:\n{db2}")

# Update weights
W1 -= learning_rate * dW1
b1 -= learning_rate * db1
W2 -= learning_rate * dW2
b2 -= learning_rate * db2

print("\n--- UPDATED WEIGHTS ---")
print(f"\nLayer 1 Weight:\n{W1}")
print(f"\nLayer 1 Bias:\n{b1}")
print(f"\nLayer 2 Weight:\n{W2}")
print(f"\nLayer 2 Bias:\n{b2}")
```
